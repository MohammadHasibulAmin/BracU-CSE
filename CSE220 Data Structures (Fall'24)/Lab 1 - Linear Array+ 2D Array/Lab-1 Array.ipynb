```python
#Task 01: Merge Lineup
import numpy as np

def mergeLineup(pokemon_1, pokemon_2):
  #To Do
  result = np.array([0]*len(pokemon_1))

  shape_1 = result.shape
  length = shape_1[0]

  for i in range(length):
    if pokemon_1[i] == None:
      result[i] = pokemon_2[length - i - 1] if pokemon_2[length - i - 1] is not None else 0
    elif pokemon_2[length-i-1] == None:
      result[i] = pokemon_1[i]
    else:
      result[i] = pokemon_1[i] + pokemon_2[length-i-1]

  return result

print("///  Task 01: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))
pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
#Task 01: Merge Lineup (Alternate)

def mergeLineup(pokemon_1, pokemon_2):
  #To Do
  result = np.array([0]*len(pokemon_1), dtype = int)

  idx1 = 0
  idx2 = pokemon_2.size-1

  while idx1 < pokemon_1.size:
    if pokemon_1[idx1] == None:
      result[idx1] = pokemon_2[idx2]
      idx1 += 1
      idx2 -= 1

    elif pokemon_2[idx2] == None:
      result[idx1] = pokemon_1[idx1]
      idx1 += 1
      idx2 -= 1

    else:
      result[idx1] = pokemon_1[idx1] + pokemon_2[idx2]
      idx1 += 1
      idx2 -= 1

  return result

print("///  Task 01: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))
pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
# Task 02: Discard Cards
import numpy as np

def discardCards(cards, t):
  # TO DO
  cards = np.array(cards)

  shape = cards.shape
  length = shape[0]

  flag = True

  count = 0
  for i in range(length):

        if cards[i] == t:
            if flag:
              cards[i] = 0
              count += 1  
              flag = False

            elif flag == False:
              flag = True

        if count == 2:
           break

  temp = np.zeros(length, dtype = int)
        
  j = 0
  for i in range(length):
     if cards[i] != 0:
        temp[j] = cards[i]
        j+= 1

  return temp

print("///  Task 02: Discard Cards  ///")
cards = np.array([1,3,7,2,5,2,2,2,0])
returned_value = discardCards(cards, 2)
print(f'Task 2: {returned_value}') # This should print [1,3,7,5,2,2,0,0,0]
unittest.output_test(returned_value, np.array([1,3,7,5,2,2,0,0,0]))

cards = np.array([5,5,5,0,0])
returned_value = discardCards(cards, 5)
print(f'Task 2: {returned_value}') # This should print [5,0,0,0,0]
unittest.output_test(returned_value, np.array([5,0,0,0,0]))
```

```python
#Task 03: Decryption Process
import numpy as np

def decrypt_matrix(matrix):
  #To Do
  shape = matrix.shape

  row_count = shape[0]
  col_count = shape[1]

  sum_array = np.zeros(col_count, dtype = int)
  k = 0

  for i in range(col_count):
    col_sum = 0
    for j in range(row_count):
      col_sum += matrix[j][i]
    sum_array[k] = col_sum
    k += 1

  result_array =np.zeros(col_count-1, dtype = int)

  r_shape = result_array.shape
  length = r_shape[0]

  for i in range(length):
    result_array[i] = sum_array[i+1] - sum_array[i]

  return result_array

matrix=np.array([[1,3,1],
                 [6,4,2],
                 [5,1,7],
                 [9,3,3],
                 [8,5,4]
                 ])

returned_array=decrypt_matrix(matrix)
print(returned_array)
#This should print [-13, 1]
```

```python
#Task 06: Matrix Compression
import numpy as np

def compress_matrix(mat):
  #TO DO
  arr = np.array(mat)

  row_count = arr.shape[0]
  col_count = arr.shape[1]

  comp = np.zeros((row_count//2,col_count//2), dtype = int)
  c_i = 0
  c_j = 0

  for row in range(0,row_count,2):
    for col in range(0,col_count,2):
      comp[c_i][c_j] = arr[row][col] + arr[row+1][col] + arr[row][col+1] + arr[row+1][col+1]
      c_j += 1
    c_i += 1
    c_j = 0

  return comp


matrix=np.array([[1,2,3,4],
                 [5,6,7,8],
                 [1,3,5,2],
                 [-2,0,6,-3]
                 ])
print_matrix(matrix)
print("...............")
returned_array=compress_matrix(matrix)
print_matrix(returned_array)
#This should print

#|  14  |  22 |
#--------------
#|  2  |  10  |
#--------------
```

```python
#Task 06: Matrix Compression (Alternate)
import numpy as np

def compress_matrix(mat):
    #TO DO
    arr = np.array(mat)
    row_count = arr.shape[0]
    col_count = arr.shape[1]

    comp = np.zeros((row_count // 2, col_count // 2), dtype=int)

    for i in range(0, row_count, 2):
        for j in range(0, col_count, 2):

            c_i, c_j = i // 2, j // 2

            comp[c_i][c_j] = (arr[i][j] + arr[i+1][j] +
                              arr[i][j+1] + arr[i+1][j+1])

    return comp


matrix=np.array([[1,2,3,4],
                 [5,6,7,8],
                 [1,3,5,2],
                 [-2,0,6,-3]
                 ])
print_matrix(matrix)
print("...............")
returned_array=compress_matrix(matrix)
print_matrix(returned_array)
#This should print

#|  14  |  22 |
#--------------
#|  2  |  10  |
#--------------
```

```python
#Task 07: Game Arena (GpT explaination)
import numpy as np

def play_game(arena):
    arr = np.array(arena)
    row_count = arr.shape[0]
    col_count = arr.shape[1]
    total = 0

    # Check cells containing numbers divisible by 50
    for row in range(row_count):
        for col in range(col_count):
            if arr[row][col] % 50 == 0 and arr[row][col] != 0:
                print(f"Checking cell with value {arr[row][col]} at ({row}, {col})")

                # Check four directions for adjacent `2`s
                if col > 0 and arr[row][col - 1] == 2:  # Left
                    total += 2
                    print(f"Left check at ({row}, {col-1}) - Total: {total}")
                if col < col_count - 1 and arr[row][col + 1] == 2:  # Right
                    total += 2
                    print(f"Right check at ({row}, {col+1}) - Total: {total}")
                if row > 0 and arr[row - 1][col] == 2:  # Up
                    total += 2
                    print(f"Up check at ({row-1}, {col}) - Total: {total}")
                if row < row_count - 1 and arr[row + 1][col] == 2:  # Down
                    total += 2
                    print(f"Down check at ({row+1}, {col}) - Total: {total}")

    # Final result based on total score
    if total > 10:
        print(f"Points Gained: {total}. Your team has survived the game")
    else:
        print(f"Points Gained: {total}. Your team is out")

# Testing with the given arenas
arena = np.array([
    [0, 2, 2, 0],
    [50, 1, 2, 0],
    [2, 2, 2, 0],
    [1, 100, 2, 0]
])
print_matrix(arena)
play_game(arena)

print(".....................")

arena = np.array([
    [0, 2, 2, 0, 2],
    [1, 50, 2, 1, 100],
    [2, 2, 2, 0, 2],
    [0, 200, 2, 0, 0]
])
print_matrix(arena)
play_game(arena)
```
