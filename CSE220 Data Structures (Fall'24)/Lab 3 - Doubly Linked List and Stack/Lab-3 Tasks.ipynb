**Assignment Part 1: Doubly Linked List**

For Assignment Part 1, you should write your full code in the following cells along with your driver codes on your own.

```python
#Assignment Part 1

class Patient:
  #write a constructor
  def __init__(self, id, name, age, bloodgroup, next, prev):
    self.id = id
    self.name = name
    self.age = age
    self.bloodgroup = bloodgroup
    self.next = next
    self.prev = prev
```

```python
class WRM:

  def __init__(self):
    #Creating the dummy head
    self.dh = Patient(None,None,None,None,None,None)
    self.dh.next = self.dh
    self.dh.prev = self.dh

  def registerPatient(self,id, name, age, bloodgroup):
    new_patient = Patient(id, name, age, bloodgroup, None, None)
    tail = self.dh.prev
    tail.next = new_patient
    new_patient.prev = tail
    new_patient.next = self.dh
    self.dh.prev = new_patient
    return "Success"

  def servePatient(self):
    served_patient = self.dh.next
    patient_name = served_patient.name
    self.dh.next = served_patient.next
    served_patient.next.prev = self.dh
    return(f"patient: {patient_name} has been served")

  def showAllPatient(self):
    temp = self.dh.next

    while temp.name != None:
      print(temp.name, end = "\n")
      temp = temp.next
    return

    while temp.name != None:
      print(temp.name, end = " ")
      temp = temp.next

  def canDoctorGoHome(self):
    if self.dh.next == self.dh:
      return "Yes"

    else: return "No"

  def cancelAll(self):
    self.dh.next = self.dh
    self.dh.prev = self.dh
    return("All patients are cancelled")

  def ReverseTheLine(self):
      if self.dh.next == self.dh:
          return "Patient list is empty"
      temp = self.dh
      while True:
          temp.next, temp.prev = temp.prev, temp.next
          temp = temp.prev
          if temp == self.dh:
                break
      return "Success"
```

```python
#Write a Tester Code in this cell
print("**Welcome to Waiting Room Management System**")

wrm = WRM()
while True:
    print("==Choose an Option==")
    print("1. Register Patient")
    print("2. Serve Patient")
    print("3. CancellAll")
    print("4. CanDoctorGoHome")
    print("5. ShowAllPatient")
    print("6. ReverseTheLine")
    print("7. Exit")

    option = int(input("Enter your choice: "))
    if option == 1:
      id = int(input("Enter Patient ID: "))
      name = input("Enter Name of the Patient: ")
      age = int(input("Enter Patient Age: "))
      bloodgroup = (input("Enter Patient Blood Group: "))
      print(wrm.registerPatient(id, name, age, bloodgroup))
    elif option == 2:
      print(wrm.servePatient())
    elif option == 3:
      print(wrm.cancelAll())
    elif option == 4:
      print(wrm.canDoctorGoHome())
    elif option == 5:
      print("Patients in waiting list:")
      wrm.showAllPatient()
    elif option == 6:
      wrm.ReverseTheLine()
    elif option == 7:
      print("Exiting Waiting Room Management System")
      break
    else:
      print("Invalid option chosen. Try again.")
```

**Assignment Part 2: Stack**

Linked List based Stack is implemented in the following cell.

```python
class Node:
  def __init__(self,elem=None,next=None):
    self.elem = elem
    self.next = next

class Stack:
  def __init__(self):
    self.__top = None

  def push(self,elem):
    nn = Node(elem,self.__top)
    self.__top = nn

  def pop(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    e = self.__top
    self.__top = self.__top.next
    return e.elem

  def peek(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    return self.__top.elem

  def isEmpty(self):
    return self.__top == None
```

Task 1: Diamond Count

```python
def diamond_count(stack,string):
  #TO DO
  count = 0
  for i in string:
        if i == "<":
          stack.push(i)
        elif i == ">":
          if stack.isEmpty() == False:
            count += 1
            stack.pop()

  return count

print('Test 01')
stack = Stack()
string = '<..><.<..>> '
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')


print('Test 02')
stack = Stack()
string = '<<<..<......<<<<....>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 1
unittest.output_test(returned_value, 1)
print('-----------------------------------------')


print('Test 03')
stack = Stack()
string = '>>><...<<..>>...>...>>>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')
```
